```python
# ---- Build full dataset (same as before, kept in Python only — NOT exported) ----
obj4_df = (
    orders
    .merge(order_item, on='order_id', how='left')
    .merge(products, on='product_id', how='left')
    .merge(order_review, on='order_id', how='left')
    .merge(customer, on='customer_id', how='left')
)

print(orders['order_status'].unique())

# Updated to match your actual values: cancelled, delivered, returned, shipped
cancel_return_statuses = ['cancelled', 'returned']
obj4_df['is_returned_or_cancelled'] = obj4_df['order_status'].isin(cancel_return_statuses).astype(int)

obj4_df = obj4_df.sort_values(['customer_id', 'order_purchase_timestamp'])
order_counts = obj4_df.groupby('customer_id')['order_id'].transform('nunique')
obj4_df['reorder_flag'] = (order_counts > 1).astype(int)
obj4_df['next_purchase_date'] = obj4_df.groupby('customer_id')['order_purchase_timestamp'].shift(-1)
obj4_df['time_to_next_purchase_days'] = (
    obj4_df['next_purchase_date'] - obj4_df['order_purchase_timestamp']
).dt.days

print(f"Full merged shape (Python only, do NOT export this): {obj4_df.shape}")

# ---------------------------------------------------------------
# SUMMARY 1: Category + Brand -> avg review score, return rate, order count
# This is what answers Question 1 — small enough for Excel
# ---------------------------------------------------------------
summary_category_brand = (
    obj4_df.groupby(['Category_name', 'brand'])
    .agg(
        avg_review_score=('review_score', 'mean'),
        return_cancel_rate=('is_returned_or_cancelled', 'mean'),
        order_count=('order_id', 'nunique')
    )
    .reset_index()
    .sort_values('avg_review_score', ascending=False)
)

print(f"Summary 1 shape: {summary_category_brand.shape}")  # tiny — safe for Excel

# ---------------------------------------------------------------
# SUMMARY 2: Customer-level -> avg review score, reorder flag, time to next purchase
# This answers Question 2 — one row per customer instead of one row per order-item
# ---------------------------------------------------------------
summary_customer = (
    obj4_df.groupby('customer_id')
    .agg(
        avg_review_score=('review_score', 'mean'),
        reorder_flag=('reorder_flag', 'max'),
        avg_time_to_next_purchase=('time_to_next_purchase_days', 'mean')
    )
    .reset_index()
)

# Bucket review scores for easier pivoting in Excel
summary_customer['review_band'] = pd.cut(
    summary_customer['avg_review_score'],
    bins=[0, 2, 3, 5],
    labels=['Low (1-2)', 'Medium (3)', 'High (4-5)']
)

print(f"Summary 2 shape: {summary_customer.shape}")  # one row per customer — much smaller

# ---------------------------------------------------------------
# Export ONLY the summaries — small files Excel can handle easily
# ---------------------------------------------------------------
with pd.ExcelWriter('objective4_summary_data.xlsx') as writer:
    summary_category_brand.to_excel(writer, sheet_name='Category_Brand_Summary', index=False)
    summary_customer.to_excel(writer, sheet_name='Customer_Reorder_Summary', index=False)

print("Exported: objective4_summary_data.xlsx (two sheets)")
```
